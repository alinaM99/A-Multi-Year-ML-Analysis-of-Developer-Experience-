```python
# ===========================================================
# TRUE Multi-Target Multi-Class (Joint Softmax, Research-Ready)
# ===========================================================
import os, numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score, balanced_accuracy_score,
    confusion_matrix, ConfusionMatrixDisplay, classification_report
)
from sklearn.utils.class_weight import compute_class_weight
from imblearn.over_sampling import SMOTE
import torch, torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from torch.cuda.amp import autocast, GradScaler
from google.colab import files
import seaborn as sns

# -----------------------------
# Config
# -----------------------------
RANDOM_STATE = 42
BATCH_SIZE   = 1024
EPOCHS       = 80
LR           = 3e-4
WEIGHT_DECAY = 1e-4
PATIENCE     = 12
DEVICE       = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

# -----------------------------
# Load original dataset
# -----------------------------
path = "devx_2024_modellingDataset_FLOtop5_union.csv"
if not os.path.exists(path):
    raise FileNotFoundError("Upload devx_2024_modellingDataset_FLOtop5_union.csv")

df = pd.read_csv(path)
print("✅ Loaded dataset:", df.shape)

TARGETS  = ["Frustration", "JobSat", "JobPerspectiveClass"]
FEATURES = [c for c in df.columns if c not in TARGETS]

# basic cleaning
df = df.fillna(df.median(numeric_only=True))
for t in TARGETS:
    df[t] = df[t].astype(int)

# -----------------------------
# Build JOINT label and balance with SMOTE
# -----------------------------
tuples = list(zip(df["Frustration"], df["JobSat"], df["JobPerspectiveClass"]))
joint_labels, uniques = pd.factorize(tuples)  # uniques is the decoder list of tuples
n_joint_classes = len(uniques)
print(f"Joint classes found: {n_joint_classes} (e.g., 3x3x3 = 27 max)")

smote = SMOTE(random_state=RANDOM_STATE, k_neighbors=3)
X_res, y_joint_res = smote.fit_resample(df[FEATURES], joint_labels)

# Decode back to separate targets (useful for per-target metrics later)
decoded = np.array([uniques[i] for i in y_joint_res])
df_bal = pd.DataFrame(X_res, columns=FEATURES)
df_bal["Frustration"]         = decoded[:, 0].astype(int)
df_bal["JobSat"]              = decoded[:, 1].astype(int)
df_bal["JobPerspectiveClass"] = decoded[:, 2].astype(int)
df_bal["joint"]               = y_joint_res.astype(int)

print("✅ After balancing:", df_bal.shape)
df_bal.to_csv("devx_balanced_multitarget_joint.csv", index=False)
print("💾 Saved:", "devx_balanced_multitarget_joint.csv")

# -----------------------------
# Train / Val / Test split (stratify on JOINT label) + keep indices
# -----------------------------
X = df_bal[FEATURES].values.astype(np.float32)
y_joint = df_bal["joint"].values.astype(np.int64)
idx_all = np.arange(len(df_bal))

X_train, X_tmp, y_train, y_tmp, idx_train, idx_tmp = train_test_split(
    X, y_joint, idx_all,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y_joint
)
X_val, X_test, y_val, y_test, idx_val, idx_test = train_test_split(
    X_tmp, y_tmp, idx_tmp,
    test_size=0.5,
    random_state=RANDOM_STATE,
    stratify=y_tmp
)

# For per-target evaluation later (still useful if needed)
dec_train = np.array([uniques[i] for i in y_train])
dec_val   = np.array([uniques[i] for i in y_val])
dec_test  = np.array([uniques[i] for i in y_test])

# -----------------------------
# Standardization
# -----------------------------
feature_scaler = StandardScaler()
X_train = feature_scaler.fit_transform(X_train)
X_val   = feature_scaler.transform(X_val)
X_test  = feature_scaler.transform(X_test)

# -----------------------------
# Dataloaders (single JOINT label)
# -----------------------------
def make_loader(X, y, bs, shuffle):
    ds = TensorDataset(torch.tensor(X, dtype=torch.float32),
                       torch.tensor(y, dtype=torch.long))
    return DataLoader(ds, batch_size=bs, shuffle=shuffle, num_workers=2, pin_memory=True)

train_loader = make_loader(X_train, y_train, BATCH_SIZE, True)
val_loader   = make_loader(X_val,   y_val,   BATCH_SIZE, False)
test_loader  = make_loader(X_test,  y_test,  BATCH_SIZE, False)

# -----------------------------
# Class weights for JOINT classes
# -----------------------------
classes = np.arange(n_joint_classes)
w_joint = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
w_joint = torch.tensor(w_joint, dtype=torch.float32, device=DEVICE)

# -----------------------------
# Model (shared MLP + single JOINT softmax head)
# -----------------------------
class ResidualMLPBlock(nn.Module):
    def __init__(self, d, drop=0.2):
        super().__init__()
        self.bn1 = nn.BatchNorm1d(d)
        self.fc1 = nn.Linear(d, 2*d)
        self.bn2 = nn.BatchNorm1d(2*d)
        self.fc2 = nn.Linear(2*d, d)
        self.drop= nn.Dropout(drop)
        self.act = nn.GELU()
    def forward(self, x):
        i = x
        x = self.bn1(x); x = self.fc1(x); x = self.act(x)
        x = self.bn2(x); x = self.drop(x); x = self.fc2(x)
        return self.act(x + i)

class JointNet(nn.Module):
    def __init__(self, in_dim, d=256, depth=5, n_classes=27, drop=0.3):
        super().__init__()
        self.stem = nn.Sequential(
            nn.Linear(in_dim, d),
            nn.BatchNorm1d(d),
            nn.GELU(),
            nn.Dropout(drop)
        )
        self.blocks = nn.Sequential(*[ResidualMLPBlock(d, drop) for _ in range(depth)])
        self.head = nn.Linear(d, n_classes)
    def forward(self, x):
        h = self.stem(x)
        h = self.blocks(h)
        return self.head(h)

model = JointNet(
    in_dim=X_train.shape[1],
    d=256,
    depth=5,
    n_classes=n_joint_classes,
    drop=0.3
).to(DEVICE)

# -----------------------------
# Optimizer / Loss / Scheduler / AMP
# -----------------------------
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2)
criterion = nn.CrossEntropyLoss(weight=w_joint)
scaler = GradScaler(enabled=(DEVICE.type == "cuda"))

# -----------------------------
# Train / Eval loops
# -----------------------------
def run_epoch(loader, train=True):
    model.train() if train else model.eval()
    tot, ys, ps = 0.0, [], []
    for xb, yb in loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        with torch.set_grad_enabled(train):
            with autocast(enabled=(DEVICE.type == "cuda")):
                logits = model(xb)
                loss = criterion(logits, yb)
            if train:
                optimizer.zero_grad(set_to_none=True)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
        tot += loss.item() * xb.size(0)
        ys.append(yb.detach().cpu().numpy())
        ps.append(logits.detach().cpu().argmax(1).numpy())
    y_true = np.concatenate(ys)
    y_pred = np.concatenate(ps)
    # Joint metrics
    acc  = accuracy_score(y_true, y_pred)
    f1   = f1_score(y_true, y_pred, average="macro")
    prec = precision_score(y_true, y_pred, average="macro")
    rec  = recall_score(y_true, y_pred, average="macro")
    bal  = balanced_accuracy_score(y_true, y_pred)
    return tot/len(loader.dataset), (acc, f1, prec, rec, bal), y_true, y_pred

best_f1, wait, best_state = -1, 0, None
for epoch in range(1, EPOCHS + 1):
    tr_loss, tr_m, _, _ = run_epoch(train_loader, True)
    va_loss, va_m, _, _ = run_epoch(val_loader, False)
    scheduler.step(epoch - 1)
    print(
        f"Epoch {epoch:03d} | TrainLoss {tr_loss:.4f} | ValLoss {va_loss:.4f} | "
        f"Val Acc {va_m[0]:.4f} | Val F1 {va_m[1]:.4f}"
    )
    if va_m[1] > best_f1 + 1e-4:
        best_f1, wait = va_m[1], 0
        best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"🛑 Early stopping at epoch {epoch}. Best Val Macro-F1 = {best_f1:.4f}")
            break

if best_state is not None:
    model.load_state_dict(best_state)

# -----------------------------
# Final JOINT evaluation
# -----------------------------
test_loss, test_m, y_true_joint, y_pred_joint = run_epoch(test_loader, False)
print("\n================= JOINT TEST METRICS =================")
print(
    f"Accuracy: {test_m[0]:.4f} | Macro-F1: {test_m[1]:.4f} | "
    f"Precision: {test_m[2]:.4f} | Recall: {test_m[3]:.4f} | BalAcc: {test_m[4]:.4f}"
)

# =====================================================
# Clean Confusion Matrix Visualization (Top 10 Joint Classes)
# =====================================================
cm = confusion_matrix(
    y_true_joint, y_pred_joint,
    labels=np.arange(len(uniques)), normalize='true'
)

unique, counts = np.unique(y_true_joint, return_counts=True)
top_idx = np.argsort(counts)[-10:]  # top 10 most frequent joint combos
top_labels = [str(uniques[i]) for i in top_idx]

cm_subset = cm[np.ix_(top_idx, top_idx)]

plt.figure(figsize=(8, 6))
sns.heatmap(
    cm_subset, annot=True, fmt=".2f", cmap="Blues",
    xticklabels=top_labels, yticklabels=top_labels,
    cbar=False, linewidths=0.5
)
plt.title("Normalized Confusion Matrix (Top 10 Joint Classes)")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

cm_df = pd.DataFrame(
    cm,
    index=[str(u) for u in uniques],
    columns=[str(u) for u in uniques]
)
cm_df.to_csv("devx_2022_confusion_full.csv", index=False)
print("💾 Saved:", "devx_confusion_full.csv")

# -----------------------------
# Derive PER-TARGET metrics from joint predictions
# -----------------------------
pred_triplets = np.array([uniques[i] for i in y_pred_joint])
true_triplets = np.array([uniques[i] for i in y_true_joint])

def per_target_report(idx, name):
    y_t = true_triplets[:, idx]
    y_p = pred_triplets[:, idx]
    print(f"\n--- {name} ---")
    print(
        f"Acc: {accuracy_score(y_t, y_p):.4f} | "
        f"Macro-F1: {f1_score(y_t, y_p, average='macro'):.4f} | "
        f"Precision: {precision_score(y_t, y_p, average='macro'):.4f} | "
        f"Recall: {recall_score(y_t, y_p, average='macro'):.4f} | "
        f"BalAcc: {balanced_accuracy_score(y_t, y_p):.4f}"
    )
    print(classification_report(y_t, y_p, digits=4))
    cm_t = confusion_matrix(y_t, y_p)
    ConfusionMatrixDisplay(cm_t).plot(cmap="Blues")
    plt.title(f"Confusion Matrix - {name}")
    plt.show()

per_target_report(0, "Frustration")
per_target_report(1, "JobSat")
per_target_report(2, "JobPerspectiveClass")

# -----------------------------
# Compact predictions (JOINT + split targets) & download
# -----------------------------
pred_df = pd.DataFrame({
    "joint_true": y_true_joint,
    "joint_pred": y_pred_joint,
    "Frustration_true": true_triplets[:, 0],
    "Frustration_pred": pred_triplets[:, 0],
    "JobSat_true": true_triplets[:, 1],
    "JobSat_pred": pred_triplets[:, 1],
    "JobPerspectiveClass_true": true_triplets[:, 2],
    "JobPerspectiveClass_pred": pred_triplets[:, 2],
})
pred_df.to_csv("devx_predictions_joint.csv", index=False)
print("💾 Saved:", "devx_predictions_joint.csv")

# ===========================================================
# Full predictions CSV (features + true + predicted targets)
# ===========================================================
full_result = df_bal.iloc[idx_test].reset_index(drop=True).copy()

full_result["Frustration_true"]         = true_triplets[:, 0]
full_result["Frustration_pred"]         = pred_triplets[:, 0]
full_result["JobSat_true"]              = true_triplets[:, 1]
full_result["JobSat_pred"]              = pred_triplets[:, 1]
full_result["JobPerspectiveClass_true"] = true_triplets[:, 2]
full_result["JobPerspectiveClass_pred"] = pred_triplets[:, 2]
full_result["Frustration_correct"] = (
    full_result["Frustration_true"] == full_result["Frustration_pred"]
)

full_result["JobSat_correct"] = (
    full_result["JobSat_true"] == full_result["JobSat_pred"]
)

full_result["JobPerspectiveClass_correct"] = (
    full_result["JobPerspectiveClass_true"] == full_result["JobPerspectiveClass_pred"]
)

# joint correctness
full_result["joint_correct"] = (
    (full_result["Frustration_correct"]) &
    (full_result["JobSat_correct"]) &
    (full_result["JobPerspectiveClass_correct"])
)


full_result.to_csv("devx_2024_full_predictions.csv", index=False)
print("💾 Saved full predictions with features:", "devx_2024_full_predictions.csv")

# -----------------------------
# Save model + scaler & download
# -----------------------------
torch.save({
    "state": model.state_dict(),
    "scaler_mean": feature_scaler.mean_,
    "scaler_scale": feature_scaler.scale_,
    "features": FEATURES,
}, "multitarget_joint_model.pth")

print("💾 Saved model as multitarget_joint_model.pth")
# ===========================================================
# ZIP ALL INTERMEDIATE + FINAL OUTPUTS
# ===========================================================
import zipfile
import os

zip_name = "devx_2024_multitarget_all_outputs.zip"

files_to_zip = [
    "devx_balanced_multitarget_joint.csv",
    "devx_confusion_full.csv",
    "devx_predictions_joint.csv",
    "devx_2024_full_predictions.csv",
    "multitarget_joint_model.pth"
]

with zipfile.ZipFile(zip_name, 'w', compression=zipfile.ZIP_DEFLATED) as z:
    for file in files_to_zip:
        if os.path.exists(file):
            z.write(file)
            print(f"✅ Added to ZIP: {file}")
        else:
            print(f"⚠️ File not found, skipped: {file}")

print("\n💾 Created ZIP:", zip_name)

from google.colab import files
files.download(zip_name)

```

```python
# ===========================================================
# TRUE Multi-Target Multi-Class (Joint Softmax, Research-Ready)
# ===========================================================
import os, numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score, balanced_accuracy_score,
    confusion_matrix, ConfusionMatrixDisplay, classification_report
)
from sklearn.utils.class_weight import compute_class_weight
from imblearn.over_sampling import SMOTE
import torch, torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from torch.cuda.amp import autocast, GradScaler
from google.colab import files
import seaborn as sns

# -----------------------------
# Config
# -----------------------------
RANDOM_STATE = 42
BATCH_SIZE   = 1024
EPOCHS       = 80
LR           = 3e-4
WEIGHT_DECAY = 1e-4
PATIENCE     = 12
DEVICE       = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print("Device:", DEVICE)

# -----------------------------
# Load original dataset
# -----------------------------
path = "devx_2025_modellingDataset_FLOtop5_union.csv"
if not os.path.exists(path):
    raise FileNotFoundError("Upload devx_2025_modellingDataset_FLOtop5_union.csv")

df = pd.read_csv(path)
print("✅ Loaded dataset:", df.shape)

TARGETS  = ["Frustration", "JobSat", "JobPerspectiveClass"]
FEATURES = [c for c in df.columns if c not in TARGETS]

# basic cleaning
df = df.fillna(df.median(numeric_only=True))
for t in TARGETS:
    df[t] = df[t].astype(int)

# -----------------------------
# Build JOINT label and balance with SMOTE
# -----------------------------
tuples = list(zip(df["Frustration"], df["JobSat"], df["JobPerspectiveClass"]))
joint_labels, uniques = pd.factorize(tuples)  # uniques is the decoder list of tuples
n_joint_classes = len(uniques)
print(f"Joint classes found: {n_joint_classes} (e.g., 3x3x3 = 27 max)")

smote = SMOTE(random_state=RANDOM_STATE, k_neighbors=3)
X_res, y_joint_res = smote.fit_resample(df[FEATURES], joint_labels)

# Decode back to separate targets (useful for per-target metrics later)
decoded = np.array([uniques[i] for i in y_joint_res])
df_bal = pd.DataFrame(X_res, columns=FEATURES)
df_bal["Frustration"]         = decoded[:, 0].astype(int)
df_bal["JobSat"]              = decoded[:, 1].astype(int)
df_bal["JobPerspectiveClass"] = decoded[:, 2].astype(int)
df_bal["joint"]               = y_joint_res.astype(int)

print("✅ After balancing:", df_bal.shape)
df_bal.to_csv("devx_balanced_multitarget_joint.csv", index=False)
print("💾 Saved:", "devx_balanced_multitarget_joint.csv")

# -----------------------------
# Train / Val / Test split (stratify on JOINT label) + keep indices
# -----------------------------
X = df_bal[FEATURES].values.astype(np.float32)
y_joint = df_bal["joint"].values.astype(np.int64)
idx_all = np.arange(len(df_bal))

X_train, X_tmp, y_train, y_tmp, idx_train, idx_tmp = train_test_split(
    X, y_joint, idx_all,
    test_size=0.2,
    random_state=RANDOM_STATE,
    stratify=y_joint
)
X_val, X_test, y_val, y_test, idx_val, idx_test = train_test_split(
    X_tmp, y_tmp, idx_tmp,
    test_size=0.5,
    random_state=RANDOM_STATE,
    stratify=y_tmp
)

# For per-target evaluation later (still useful if needed)
dec_train = np.array([uniques[i] for i in y_train])
dec_val   = np.array([uniques[i] for i in y_val])
dec_test  = np.array([uniques[i] for i in y_test])

# -----------------------------
# Standardization
# -----------------------------
feature_scaler = StandardScaler()
X_train = feature_scaler.fit_transform(X_train)
X_val   = feature_scaler.transform(X_val)
X_test  = feature_scaler.transform(X_test)

# -----------------------------
# Dataloaders (single JOINT label)
# -----------------------------
def make_loader(X, y, bs, shuffle):
    ds = TensorDataset(torch.tensor(X, dtype=torch.float32),
                       torch.tensor(y, dtype=torch.long))
    return DataLoader(ds, batch_size=bs, shuffle=shuffle, num_workers=2, pin_memory=True)

train_loader = make_loader(X_train, y_train, BATCH_SIZE, True)
val_loader   = make_loader(X_val,   y_val,   BATCH_SIZE, False)
test_loader  = make_loader(X_test,  y_test,  BATCH_SIZE, False)

# -----------------------------
# Class weights for JOINT classes
# -----------------------------
classes = np.arange(n_joint_classes)
w_joint = compute_class_weight(class_weight='balanced', classes=classes, y=y_train)
w_joint = torch.tensor(w_joint, dtype=torch.float32, device=DEVICE)

# -----------------------------
# Model (shared MLP + single JOINT softmax head)
# -----------------------------
class ResidualMLPBlock(nn.Module):
    def __init__(self, d, drop=0.2):
        super().__init__()
        self.bn1 = nn.BatchNorm1d(d)
        self.fc1 = nn.Linear(d, 2*d)
        self.bn2 = nn.BatchNorm1d(2*d)
        self.fc2 = nn.Linear(2*d, d)
        self.drop= nn.Dropout(drop)
        self.act = nn.GELU()
    def forward(self, x):
        i = x
        x = self.bn1(x); x = self.fc1(x); x = self.act(x)
        x = self.bn2(x); x = self.drop(x); x = self.fc2(x)
        return self.act(x + i)

class JointNet(nn.Module):
    def __init__(self, in_dim, d=256, depth=5, n_classes=27, drop=0.3):
        super().__init__()
        self.stem = nn.Sequential(
            nn.Linear(in_dim, d),
            nn.BatchNorm1d(d),
            nn.GELU(),
            nn.Dropout(drop)
        )
        self.blocks = nn.Sequential(*[ResidualMLPBlock(d, drop) for _ in range(depth)])
        self.head = nn.Linear(d, n_classes)
    def forward(self, x):
        h = self.stem(x)
        h = self.blocks(h)
        return self.head(h)

model = JointNet(
    in_dim=X_train.shape[1],
    d=256,
    depth=5,
    n_classes=n_joint_classes,
    drop=0.3
).to(DEVICE)

# -----------------------------
# Optimizer / Loss / Scheduler / AMP
# -----------------------------
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10, T_mult=2)
criterion = nn.CrossEntropyLoss(weight=w_joint)
scaler = GradScaler(enabled=(DEVICE.type == "cuda"))

# -----------------------------
# Train / Eval loops
# -----------------------------
def run_epoch(loader, train=True):
    model.train() if train else model.eval()
    tot, ys, ps = 0.0, [], []
    for xb, yb in loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        with torch.set_grad_enabled(train):
            with autocast(enabled=(DEVICE.type == "cuda")):
                logits = model(xb)
                loss = criterion(logits, yb)
            if train:
                optimizer.zero_grad(set_to_none=True)
                scaler.scale(loss).backward()
                scaler.step(optimizer)
                scaler.update()
        tot += loss.item() * xb.size(0)
        ys.append(yb.detach().cpu().numpy())
        ps.append(logits.detach().cpu().argmax(1).numpy())
    y_true = np.concatenate(ys)
    y_pred = np.concatenate(ps)
    # Joint metrics
    acc  = accuracy_score(y_true, y_pred)
    f1   = f1_score(y_true, y_pred, average="macro")
    prec = precision_score(y_true, y_pred, average="macro")
    rec  = recall_score(y_true, y_pred, average="macro")
    bal  = balanced_accuracy_score(y_true, y_pred)
    return tot/len(loader.dataset), (acc, f1, prec, rec, bal), y_true, y_pred

best_f1, wait, best_state = -1, 0, None
for epoch in range(1, EPOCHS + 1):
    tr_loss, tr_m, _, _ = run_epoch(train_loader, True)
    va_loss, va_m, _, _ = run_epoch(val_loader, False)
    scheduler.step(epoch - 1)
    print(
        f"Epoch {epoch:03d} | TrainLoss {tr_loss:.4f} | ValLoss {va_loss:.4f} | "
        f"Val Acc {va_m[0]:.4f} | Val F1 {va_m[1]:.4f}"
    )
    if va_m[1] > best_f1 + 1e-4:
        best_f1, wait = va_m[1], 0
        best_state = {k: v.cpu().clone() for k, v in model.state_dict().items()}
    else:
        wait += 1
        if wait >= PATIENCE:
            print(f"🛑 Early stopping at epoch {epoch}. Best Val Macro-F1 = {best_f1:.4f}")
            break

if best_state is not None:
    model.load_state_dict(best_state)

# -----------------------------
# Final JOINT evaluation
# -----------------------------
test_loss, test_m, y_true_joint, y_pred_joint = run_epoch(test_loader, False)
print("\n================= JOINT TEST METRICS =================")
print(
    f"Accuracy: {test_m[0]:.4f} | Macro-F1: {test_m[1]:.4f} | "
    f"Precision: {test_m[2]:.4f} | Recall: {test_m[3]:.4f} | BalAcc: {test_m[4]:.4f}"
)

# =====================================================
# Clean Confusion Matrix Visualization (Top 10 Joint Classes)
# =====================================================
cm = confusion_matrix(
    y_true_joint, y_pred_joint,
    labels=np.arange(len(uniques)), normalize='true'
)

unique, counts = np.unique(y_true_joint, return_counts=True)
top_idx = np.argsort(counts)[-10:]  # top 10 most frequent joint combos
top_labels = [str(uniques[i]) for i in top_idx]

cm_subset = cm[np.ix_(top_idx, top_idx)]

plt.figure(figsize=(8, 6))
sns.heatmap(
    cm_subset, annot=True, fmt=".2f", cmap="Blues",
    xticklabels=top_labels, yticklabels=top_labels,
    cbar=False, linewidths=0.5
)
plt.title("Normalized Confusion Matrix (Top 10 Joint Classes)")
plt.xlabel("Predicted Label")
plt.ylabel("True Label")
plt.xticks(rotation=45, ha='right')
plt.yticks(rotation=0)
plt.tight_layout()
plt.show()

cm_df = pd.DataFrame(
    cm,
    index=[str(u) for u in uniques],
    columns=[str(u) for u in uniques]
)
cm_df.to_csv("devx_confusion_full.csv", index=True)
print("💾 Saved:", "devx_confusion_full.csv")

# -----------------------------
# Derive PER-TARGET metrics from joint predictions
# -----------------------------
pred_triplets = np.array([uniques[i] for i in y_pred_joint])
true_triplets = np.array([uniques[i] for i in y_true_joint])

def per_target_report(idx, name):
    y_t = true_triplets[:, idx]
    y_p = pred_triplets[:, idx]
    print(f"\n--- {name} ---")
    print(
        f"Acc: {accuracy_score(y_t, y_p):.4f} | "
        f"Macro-F1: {f1_score(y_t, y_p, average='macro'):.4f} | "
        f"Precision: {precision_score(y_t, y_p, average='macro'):.4f} | "
        f"Recall: {recall_score(y_t, y_p, average='macro'):.4f} | "
        f"BalAcc: {balanced_accuracy_score(y_t, y_p):.4f}"
    )
    print(classification_report(y_t, y_p, digits=4))
    cm_t = confusion_matrix(y_t, y_p)
    ConfusionMatrixDisplay(cm_t).plot(cmap="Blues")
    plt.title(f"Confusion Matrix - {name}")
    plt.show()

per_target_report(0, "Frustration")
per_target_report(1, "JobSat")
per_target_report(2, "JobPerspectiveClass")

# -----------------------------
# Compact predictions (JOINT + split targets) & download
# -----------------------------
pred_df = pd.DataFrame({
    "joint_true": y_true_joint,
    "joint_pred": y_pred_joint,
    "Frustration_true": true_triplets[:, 0],
    "Frustration_pred": pred_triplets[:, 0],
    "JobSat_true": true_triplets[:, 1],
    "JobSat_pred": pred_triplets[:, 1],
    "JobPerspectiveClass_true": true_triplets[:, 2],
    "JobPerspectiveClass_pred": pred_triplets[:, 2],
})
pred_df.to_csv("devx_predictions_joint.csv", index=False)
print("💾 Saved:", "devx_predictions_joint.csv")
# ===========================================================
# Full predictions CSV (features + true + predicted targets)
# ===========================================================
full_result = df_bal.iloc[idx_test].reset_index(drop=True).copy()

full_result["Frustration_true"]         = true_triplets[:, 0]
full_result["Frustration_pred"]         = pred_triplets[:, 0]
full_result["JobSat_true"]              = true_triplets[:, 1]
full_result["JobSat_pred"]              = pred_triplets[:, 1]
full_result["JobPerspectiveClass_true"] = true_triplets[:, 2]
full_result["JobPerspectiveClass_pred"] = pred_triplets[:, 2]

full_result["Frustration_correct"] = (
    full_result["Frustration_true"] == full_result["Frustration_pred"]
)

full_result["JobSat_correct"] = (
    full_result["JobSat_true"] == full_result["JobSat_pred"]
)

full_result["JobPerspectiveClass_correct"] = (
    full_result["JobPerspectiveClass_true"] == full_result["JobPerspectiveClass_pred"]
)

full_result["joint_correct"] = (
    (full_result["Frustration_correct"]) &
    (full_result["JobSat_correct"]) &
    (full_result["JobPerspectiveClass_correct"])
)


full_result.to_csv("devx_2025_full_predictions.csv", index=False)
print("💾 Saved full predictions with features:", "devx_2025_full_predictions.csv")

# -----------------------------
# Save model + scaler & download
# -----------------------------
torch.save({
    "state": model.state_dict(),
    "scaler_mean": feature_scaler.mean_,
    "scaler_scale": feature_scaler.scale_,
    "features": FEATURES,
}, "multitarget_joint_model.pth")

print("💾 Saved model as multitarget_joint_model.pth")

# ===========================================================
# ZIP ALL INTERMEDIATE + FINAL OUTPUTS
# ===========================================================
import zipfile
import os

zip_name = "devx_2025_multitarget_all_outputs.zip"

files_to_zip = [
    "devx_balanced_multitarget_joint.csv",
    "devx_confusion_full.csv",
    "devx_predictions_joint.csv",
    "devx_2025_full_predictions.csv",
    "multitarget_joint_model.pth"
]

with zipfile.ZipFile(zip_name, 'w', compression=zipfile.ZIP_DEFLATED) as z:
    for file in files_to_zip:
        if os.path.exists(file):
            z.write(file)
            print(f"✅ Added to ZIP: {file}")
        else:
            print(f"⚠️ File not found, skipped: {file}")

print("\n💾 Created ZIP:", zip_name)

from google.colab import files
files.download(zip_name)


```

```python
# ===========================================================
# TRUE Multi-Target Multi-Class (Joint Softmax, 2022 DevX Version)
# ===========================================================
import os, numpy as np, pandas as pd, matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.metrics import (
    accuracy_score, f1_score, precision_score, recall_score,
    balanced_accuracy_score, confusion_matrix,
    ConfusionMatrixDisplay, classification_report
)
from sklearn.utils.class_weight import compute_class_weight
from imblearn.over_sampling import SMOTE
import torch, torch.nn as nn
from torch.utils.data import TensorDataset, DataLoader
from torch.cuda.amp import autocast, GradScaler
import seaborn as sns
from collections import Counter

# -----------------------------
# Config
# -----------------------------
RANDOM_STATE = 42
BATCH_SIZE   = 1024
EPOCHS       = 80
LR           = 3e-4
WEIGHT_DECAY = 1e-4
PATIENCE     = 12
DEVICE       = torch.device("cuda" if torch.cuda.is_available() else "cpu")

print("Device:", DEVICE)

# -----------------------------
# Load dataset
# -----------------------------
path = "devx_2022_modellingDataset_FLOtop5_union.csv"
df = pd.read_csv(path)
print("✅ Loaded:", df.shape)

TARGETS = ["MentalHealth", "BuyNewTool", "PurchaseInfluence"]
FEATURES = [c for c in df.columns if c not in TARGETS]

df = df.fillna(df.median(numeric_only=True))
for t in TARGETS:
    df[t] = df[t].astype(int)

# ===========================================================
# 1) Build joint labels from TRIPLETS
# ===========================================================
triplets = list(zip(df["MentalHealth"], df["BuyNewTool"], df["PurchaseInfluence"]))
joint_labels, uniques = pd.factorize(pd.Series(triplets))
uniques = np.array(uniques)  # list of actual triplets
print("Initial joint classes:", len(uniques))

# ===========================================================
# 2) Remove rare joint classes (<5 samples)
# ===========================================================
counts = Counter(joint_labels)
valid = {cls for cls, n in counts.items() if n >= 5}
keep_idx = [i for i, lbl in enumerate(joint_labels) if lbl in valid]

print("Removed", len(df) - len(keep_idx), "rare-sample rows.")

df = df.iloc[keep_idx].reset_index(drop=True)
joint_labels = np.array(joint_labels)[keep_idx]

# ===========================================================
# 3) RE-FACTORIZING correctly using TRIPLETS after filtering
# ===========================================================
triplets_filtered = list(zip(df["MentalHealth"], df["BuyNewTool"], df["PurchaseInfluence"]))
joint_labels, uniques = pd.factorize(pd.Series(triplets_filtered))
uniques = np.array(uniques)  # proper shape: (n_classes, 3)
n_joint_classes = len(uniques)

print("Joint classes AFTER filtering:", n_joint_classes)

# ===========================================================
# 4) SMOTE Oversampling
# ===========================================================
smote = SMOTE(random_state=RANDOM_STATE, k_neighbors=3)
X_res, y_joint_res = smote.fit_resample(df[FEATURES], joint_labels)

decoded = np.array([uniques[i] for i in y_joint_res])  # shape (N, 3)

df_bal = pd.DataFrame(X_res, columns=FEATURES)
df_bal["MentalHealth"]        = decoded[:,0]
df_bal["BuyNewTool"]          = decoded[:,1]
df_bal["PurchaseInfluence"]   = decoded[:,2]
df_bal["joint"]               = y_joint_res
df_bal.to_csv("devx_2022_balanced_multitarget_joint.csv", index=False)
print("💾 Saved balanced dataset")

print("Balanced dataset:", df_bal.shape)

# ===========================================================
# 5) Train/Val/Test split
# ===========================================================
X = df_bal[FEATURES].values.astype(np.float32)
y = df_bal["joint"].values.astype(np.int64)

# full-row feature copy for final CSV
full_features_df = df_bal[FEATURES].copy()

X_train, X_tmp, y_train, y_tmp, idx_train, idx_tmp = train_test_split(
    X, y, np.arange(len(df_bal)), test_size=0.2,
    random_state=RANDOM_STATE, stratify=y
)

X_val, X_test, y_val, y_test, idx_val, idx_test = train_test_split(
    X_tmp, y_tmp, idx_tmp, test_size=0.5,
    random_state=RANDOM_STATE, stratify=y_tmp
)

# Decode triplets for evaluation
true_trip_train = np.array([uniques[i] for i in y_train])
true_trip_val   = np.array([uniques[i] for i in y_val])
true_trip_test  = np.array([uniques[i] for i in y_test])

# ===========================================================
# 6) Standardization
# ===========================================================
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_val   = scaler.transform(X_val)
X_test  = scaler.transform(X_test)

# ===========================================================
# 7) Dataloaders
# ===========================================================
def make_loader(X, y, batch, shuffle):
    ds = TensorDataset(torch.tensor(X, dtype=torch.float32),
                       torch.tensor(y, dtype=torch.long))
    return DataLoader(ds, batch_size=batch, shuffle=shuffle)

train_loader = make_loader(X_train, y_train, BATCH_SIZE, True)
val_loader   = make_loader(X_val,   y_val,   BATCH_SIZE, False)
test_loader  = make_loader(X_test,  y_test,  BATCH_SIZE, False)

# ===========================================================
# 8) Class Weights
# ===========================================================
w_joint = compute_class_weight("balanced", classes=np.arange(n_joint_classes), y=y_train)
w_joint = torch.tensor(w_joint, dtype=torch.float32, device=DEVICE)

# ===========================================================
# 9) Model Definition (Residual MLP)
# ===========================================================
class ResidualMLPBlock(nn.Module):
    def __init__(self, d, drop=0.25):
        super().__init__()
        self.bn1 = nn.BatchNorm1d(d)
        self.fc1 = nn.Linear(d, 2*d)
        self.bn2 = nn.BatchNorm1d(2*d)
        self.fc2 = nn.Linear(2*d, d)
        self.drop= nn.Dropout(drop)
        self.act = nn.GELU()
    def forward(self, x):
        skip = x
        x = self.bn1(x); x = self.act(self.fc1(x))
        x = self.bn2(x); x = self.drop(x); x = self.fc2(x)
        return self.act(x + skip)

class JointNet(nn.Module):
    def __init__(self, in_dim, d=256, depth=5, n_classes=30):
        super().__init__()
        self.stem = nn.Sequential(
            nn.Linear(in_dim, d),
            nn.BatchNorm1d(d),
            nn.GELU(),
            nn.Dropout(0.3)
        )
        self.blocks = nn.Sequential(*[ResidualMLPBlock(d) for _ in range(depth)])
        self.head = nn.Linear(d, n_classes)
    def forward(self, x):
        x = self.stem(x)
        x = self.blocks(x)
        return self.head(x)

model = JointNet(in_dim=X_train.shape[1], d=256, depth=5, n_classes=n_joint_classes).to(DEVICE)

# ===========================================================
# 10) Optimizer, Loss, AMP, Scheduler
# ===========================================================
criterion = nn.CrossEntropyLoss(weight=w_joint)
optimizer = torch.optim.AdamW(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)
scheduler = torch.optim.lr_scheduler.CosineAnnealingWarmRestarts(optimizer, T_0=10)
gscaler = GradScaler(enabled=(DEVICE.type=="cuda"))  # <-- renamed to avoid conflict

# ===========================================================
# 11) Training Loop
# ===========================================================
def run_epoch(loader, train=True):
    model.train() if train else model.eval()
    total_loss, ys, ps = 0.0, [], []
    for xb, yb in loader:
        xb, yb = xb.to(DEVICE), yb.to(DEVICE)
        with torch.set_grad_enabled(train):
            with autocast(enabled=(DEVICE.type=="cuda")):
                logits = model(xb)
                loss = criterion(logits, yb)
            if train:
                optimizer.zero_grad()
                gscaler.scale(loss).backward()
                gscaler.step(optimizer)
                gscaler.update()
        total_loss += loss.item() * xb.size(0)
        ys.append(yb.cpu().numpy())
        ps.append(logits.argmax(1).cpu().numpy())
    y_true = np.concatenate(ys)
    y_pred = np.concatenate(ps)
    return (
        total_loss / len(loader.dataset),
        accuracy_score(y_true, y_pred),
        f1_score(y_true, y_pred, average="macro"),
        precision_score(y_true, y_pred, average="macro", zero_division=0),
        recall_score(y_true, y_pred, average="macro", zero_division=0),
        balanced_accuracy_score(y_true, y_pred),
        y_true, y_pred
    )

best_f1 = -1
pat = 0
best_state = None

for epoch in range(1, EPOCHS + 1):
    tr_loss, tr_acc, tr_f1, tr_prec, tr_rec, _, _, _ = run_epoch(train_loader, True)
    va_loss, va_acc, va_f1, va_prec, va_rec, va_bal, _, _ = run_epoch(val_loader, False)
    scheduler.step(epoch)

    print(f"E{epoch:03d} | TrainLoss {tr_loss:.4f} | ValLoss {va_loss:.4f} "
          f"| ValAcc {va_acc:.4f} | ValF1 {va_f1:.4f}")

    if va_f1 > best_f1 + 1e-4:
        best_f1 = va_f1
        best_state = {k: v.cpu().clone() for k,v in model.state_dict().items()}
        pat = 0
    else:
        pat += 1
        if pat >= PATIENCE:
            print("🛑 Early stopping.")
            break

if best_state:
    model.load_state_dict(best_state)

# ===========================================================
# 12) Final Test Evaluation
# ===========================================================
tl, acc, f1, prec, rec, bal, y_true_joint, y_pred_joint = run_epoch(test_loader, False)

print("\n=== JOINT TEST RESULTS ===")
print(f"Accuracy: {acc:.4f} | Macro-F1: {f1:.4f} | "
      f"Precision: {prec:.4f} | Recall: {rec:.4f} | Balanced Acc: {bal:.4f}")

# decode
true_trip = np.array([uniques[i] for i in y_true_joint])
pred_trip = np.array([uniques[i] for i in y_pred_joint])

cm = confusion_matrix(y_true_joint, y_pred_joint)

cm_df = pd.DataFrame(
    cm,
    index=[str(u) for u in uniques],
    columns=[str(u) for u in uniques]
)

cm_df.to_csv("devx_2022_confusion_full.csv")
print("💾 Saved confusion matrix")

pred_df = pd.DataFrame({
    "joint_true": y_true_joint,
    "joint_pred": y_pred_joint,
    "MH_true": true_trip[:,0],
    "MH_pred": pred_trip[:,0],
    "BT_true": true_trip[:,1],
    "BT_pred": pred_trip[:,1],
    "PI_true": true_trip[:,2],
    "PI_pred": pred_trip[:,2],
})

pred_df.to_csv("devx_2022_predictions_joint.csv", index=False)
print("💾 Saved compact predictions")

# ===========================================================
# 13) Per-target metrics
# ===========================================================
def report_target(idx, name):
    yt = true_trip[:, idx]
    yp = pred_trip[:, idx]
    print(f"\n--- {name} ---")
    print(classification_report(yt, yp, digits=3))
    ConfusionMatrixDisplay(confusion_matrix(yt, yp)).plot(cmap="Blues")
    plt.title(name)
    plt.show()

report_target(0, "MentalHealth")
report_target(1, "BuyNewTool")
report_target(2, "PurchaseInfluence")

# ===========================================================
# 14) Save full predictions CSV (with features)
# ===========================================================
result = full_features_df.iloc[idx_test].copy()
result["MH_true"]  = true_trip[:,0]
result["MH_pred"]  = pred_trip[:,0]
result["BT_true"]  = true_trip[:,1]
result["BT_pred"]  = pred_trip[:,1]
result["PI_true"]  = true_trip[:,2]
result["PI_pred"]  = pred_trip[:,2]

result["MH_correct"] = (
    result["MH_true"] == result["MH_pred"]
)

result["BT_correct"] = (
    result["BT_true"] == result["BT_pred"]
)

result["PI_correct"] = (
    result["PI_true"] == result["PI_pred"]
)

result["joint_correct"] = (
    result["MH_correct"] &
    result["BT_correct"] &
    result["PI_correct"]
)


result.to_csv("devx_2022_full_predictions.csv", index=False)
print("💾 Saved predictions:", "devx_2022_full_predictions.csv")

# ===========================================================
# 15) Save model + scaler + decoder
# ===========================================================
torch.save({
    "state_dict": model.state_dict(),
    "scaler_mean": scaler.mean_,
    "scaler_scale": scaler.scale_,
    "features": FEATURES,
    "combos": uniques.tolist(),  # decoder for joint → (MH, BT, PI)
}, "devx_2022_joint_model.pth")

print("💾 Saved model as devx_2022_joint_model.pth")

import zipfile, os

zip_name = "devx_2022_multitarget_outputs.zip"

files_to_zip = [
    "devx_2022_balanced_multitarget_joint.csv",
    "devx_2022_confusion_full.csv",
    "devx_2022_predictions_joint.csv",
    "devx_2022_full_predictions.csv",
    "devx_2022_joint_model.pth"
]


with zipfile.ZipFile(zip_name, 'w', compression=zipfile.ZIP_DEFLATED) as z:
    for f in files_to_zip:
        if os.path.exists(f):
            z.write(f)

print("💾 Created ZIP:", zip_name)

```
