```python
import pandas as pd
import numpy as np
import re

# ===============================================================
# LOAD + STRIP SPACES
# ===============================================================
df = pd.read_csv("survey_results_public.csv", low_memory=False)
df = df.applymap(lambda x: x.strip() if isinstance(x, str) else x)
print("🔥 Loaded dataset")


# ===============================================================
# DROP COLUMNS DROPPED IN 2025 + ADDITIONAL 2022 DROPS
# ===============================================================
drop_cols = [
    # WANT-TO-WORK-WITH (same removal logic as 2025)
    "LanguageWantToWorkWith", "DatabaseWantToWorkWith",
    "PlatformWantToWorkWith", "WebframeWantToWorkWith",
    "MiscTechWantToWorkWith", "ToolsTechWantToWorkWith",
    "OfficeStackAsyncWantToWorkWith", "OfficeStackSyncWantToWorkWith",
    "NEWCollabToolsWantToWorkWith",

    # 2025 DROPPED COLUMNS (general)
    "DevEnvsWantToWorkWith", "SOTagsWantToWorkWith",
    "OfficeStackWant Entry", "CommPlatformWantToWorkWith",
    "AIModelsWantToWorkWith", "AISent", "AIAcc",
    "AIToolCurrently partially AI", "AIToolDon't plan to use AI for this task",
    "AIToolPlan to partially use AI", "AIToolPlan to mostly use AI",
    "AIToolCurrently mostly AI",

    # EMPTY / USELESS 2022 columns
    "VCHostingPersonal use", "VCHostingProfessional use",

    # REMOVE DEMOGRAPHICS / COMPENSATION
    "Country", "Currency", "CompTotal", "ConvertedCompYearly",
    "Sexuality", "Ethnicity", "SurveyLength", "SurveyEase",
    "Gender", "Trans", "CompFreq","SOAccount","Accessibility",

    # OTHER NON-FACTOR ID
    "ResponseId", "Blockchain"
]

df.drop(columns=[c for c in drop_cols if c in df.columns], inplace=True)
print("🧹 Dropped irrelevant columns")


# ===============================================================
# MULTI-SELECT → COUNT (GLOBAL RULE) — FIXED
# ===============================================================

# Do NOT convert these into counts — they need raw text!
EXCLUDE_MS = ["MentalHealth"]

def is_multiselect(col):
    try:
        s = df[col].dropna().astype(str)
        return (s.str.contains(";").mean() > 0.05)
    except:
        return False

multi_cols = [
    c for c in df.columns
    if c not in EXCLUDE_MS and is_multiselect(c)
]

for col in multi_cols:
    df[col] = df[col].apply(lambda x: len(str(x).split(";")) if pd.notna(x) else 0)

print(f"✔ Converted {len(multi_cols)} multi-select columns into counts (EXCLUDING MentalHealth)")



# ===============================================================
# TARGET 1 — MentalHealth (OVERWRITE ORIGINAL, FIXED NORMALIZER)
# ===============================================================

import re
import unicodedata

def normalize_text(t):
    if not isinstance(t, str):
        return ""
    t = t.lower().strip()

    # Normalize unicode accents and punctuation
    t = unicodedata.normalize("NFKD", t)

    # Replace fancy quotes
    replacements = {
        "’": "'", "‘": "'", "“": '"', "”": '"',
        "â€™": "'", "â€˜": "'", "â€œ": '"', "â€": '"'
    }
    for bad, good in replacements.items():
        t = t.replace(bad, good)

    # Remove double spaces / weird whitespace
    t = re.sub(r"\s+", " ", t)

    return t.strip()


MENTAL_VALID = {
    normalize_text("I have a concentration and/or memory disorder (e.g., ADHD, etc.)"),
    normalize_text("I have a mood or emotional disorder (e.g., depression, bipolar disorder, etc.)"),
    normalize_text("I have an anxiety disorder"),
    normalize_text("I have autism / an autism spectrum disorder (e.g. Asperger's, etc.)"),
    normalize_text("I have learning differences (e.g., Dyslexic, Dyslexia, etc.)")
}

def encode_mental(entry):
    if pd.isna(entry):
        return 0

    parts = [normalize_text(x) for x in str(entry).split(";")]

    count = sum(1 for x in parts if x in MENTAL_VALID)

    if count == 0:
        return 0
    elif count <= 2:
        return 1
    else:
        return 2

df["MentalHealth"] = df["MentalHealth"].apply(encode_mental).astype("Int64")

print("✔ MentalHealth fixed → 0/1/2")
print(df["MentalHealth"].value_counts())



# ===============================================================
# TARGET 2 — BuyNewTool (OVERWRITE ORIGINAL)
# ===============================================================

def encode_buy(n):
    if pd.isna(n): return 0
    n = int(n)
    if n == 0: return 0
    elif n <= 2: return 1
    return 2

df["BuyNewTool"] = df["BuyNewTool"].apply(encode_buy).astype("Int64")
print("✔ BuyNewTool → 0/1/2")


# ===============================================================
# TARGET 3 — PurchaseInfluence (OVERWRITE ORIGINAL)
# ===============================================================
purchase_map = {
    "I have little or no influence": 0,
    "I have some influence": 1,
    "I have a great deal of influence": 2
}

df["PurchaseInfluence"] = df["PurchaseInfluence"].map(purchase_map).astype("Int64")
print("✔ PurchaseInfluence → 0/1/2")


# ===============================================================
# ORDINAL ENCODING (MATCHING 2025 LOGIC)
# ===============================================================

def map_mainbranch(s):
    if pd.isna(s): return pd.NA
    s = s.lower()
    if "hobby" in s: return 0
    if "learning" in s: return 1
    if "not primarily" in s: return 2
    if "developer by profession" in s: return 3
    return pd.NA

df["MainBranch"] = df["MainBranch"].apply(map_mainbranch).astype("Int64")

remote_map = {
    "Full in-person": 0,
    "Hybrid (some remote, some in-person)": 1,
    "Fully remote": 2
}
df["RemoteWork"] = df["RemoteWork"].map(remote_map).astype("Int64")

org_map = {
    "Just me - I am a freelancer, sole proprietor, etc.": 0,
    "2 to 9 employees": 1,
    "10 to 19 employees": 2,
    "20 to 99 employees": 3,
    "100 to 499 employees": 4,
    "500 to 999 employees": 5,
    "1,000 to 4,999 employees": 6,
    "5,000 to 9,999 employees": 7,
    "10,000 or more employees": 8
}
df["OrgSize"] = df["OrgSize"].map(org_map).astype("Int64")

# SOVisitFreq / SOPartFreq
visit_map = {
    "Less than once per month or monthly": 1,
    "A few times per month or weekly": 2,
    "A few times per week": 3,
    "Daily or almost daily": 4,
    "Multiple times per day": 5
}
df["SOVisitFreq"] = df["SOVisitFreq"].map(visit_map).astype("Int64")

part_map = {
    "I have never participated in Q&A on Stack Overflow": 0,
    "Less than once per month or monthly": 1,
    "A few times per month or weekly": 2,
    "A few times per week": 3,
    "Daily or almost daily": 4,
    "Multiple times per day": 5
}
df["SOPartFreq"] = df["SOPartFreq"].map(part_map).astype("Int64")

print("✔ SO variables encoded like 2025")


# ===============================================================
# CLEAN YEAR FIELDS (YearsCode, YearsCodePro, WorkExp)
# ===============================================================

def coerce_years(v):
    if pd.isna(v): return np.nan
    s = str(v).lower().strip()
    nums = re.findall(r"\d+\.?\d*", s)
    if nums:
        n = float(nums[0])
        if n > 50: return 51
        if n < 1: return 0.5
        return n
    if "less" in s: return 0.5
    if "few" in s or "some" in s: return 1
    if "more" in s: return 51
    if "decade" in s: return 10
    if "none" in s or "never" in s: return 0
    return np.nan

for col in ["YearsCode", "YearsCodePro", "WorkExp"]:
    if col in df.columns:
        df[col] = df[col].apply(coerce_years)

print("✔ Cleaned year experience columns")


# ===============================================================
# SIMPLE LOW-CARDINALITY CATEGORICAL ENCODING (safe)
# ===============================================================
cat_cols = [c for c in df.columns if df[c].dtype == object and df[c].nunique() < 20]

for col in cat_cols:
    df[col] = df[col].astype("category").cat.codes.replace(-1, np.nan)

print("✔ Encoded low-cardinality categorical columns")


# ===============================================================
# REPLACE -1 WITH NaN (cleanup)
# ===============================================================
df.replace(-1, np.nan, inplace=True)


```

```python
# ===============================================================
# FINAL STEP — MICE IMPUTATION + ROUNDING + TARGET CLIPPING (2022)
# ===============================================================

from sklearn.experimental import enable_iterative_imputer
from sklearn.impute import IterativeImputer
from sklearn.linear_model import BayesianRidge
import numpy as np

# ---------------------------------------------------------------
# 1. Copy cleaned dataset
# ---------------------------------------------------------------
df_clean = df.copy()

print("Before MICE: total missing =", df_clean.isna().sum().sum())

# ---------------------------------------------------------------
# 2. SELECT ONLY NUMERIC COLUMNS
# ---------------------------------------------------------------
numeric_cols = df_clean.select_dtypes(
    include=["float64", "int64", "Int64"]
).columns

X_numeric = df_clean[numeric_cols]

# ---------------------------------------------------------------
# 3. CREATE & RUN MICE
# ---------------------------------------------------------------
mice = IterativeImputer(
    estimator=BayesianRidge(),
    max_iter=20,
    sample_posterior=False,
    random_state=42
)

X_imputed = mice.fit_transform(X_numeric)

# ---------------------------------------------------------------
# 4. REPLACE BACK INTO DATAFRAME
# ---------------------------------------------------------------
df_clean[numeric_cols] = X_imputed

# ---------------------------------------------------------------
# 5. ROUND ALL NUMERIC COLUMNS (MICE produces decimals)
# ---------------------------------------------------------------
df_clean[numeric_cols] = df_clean[numeric_cols].round()

# ---------------------------------------------------------------
# 6. CLIP TARGET VARIABLES TO 0–2
# ---------------------------------------------------------------
target_cols_2022 = [
    "MentalHealth",
    "BuyNewTool",
    "PurchaseInfluence"
]

for col in target_cols_2022:
    if col in df_clean.columns:
        df_clean[col] = df_clean[col].clip(0, 2).astype("Int64")

print("After MICE + rounding + clipping, total missing =", df_clean.isna().sum().sum())
# ---------------------------------------------------------------
# 6.1 RENAME VARIABLES
# ---------------------------------------------------------------

rename_map_2022 = {

    # ---- Work Behaviour ----
    "TimeSearching": "search_time",
    "TimeAnswering": "answer_time",

    # ---- Knowledge / Workplace ----
    "Knowledge_1": "cross_team_interaction",
    "Knowledge_2": "knowledge_silos",
    "Knowledge_3": "info_accessibility",
    "Knowledge_4": "answer_findability",
    "Knowledge_5": "resource_awareness",
    "Knowledge_6": "repeated_questions",
    "Knowledge_7": "workflow_interruptions",

    # ---- Help Dynamics ----
    "Frequency_1": "help_frequency",
    "Frequency_2": "cross_team_help_frequency",
    "Frequency_3": "knowledge_silo_frequency",

    # ---- Organization ----
    "OrgSize": "org_size",
    "DevType": "developer_role",
    "WorkExp": "work_experience",
    "YearsCodePro": "coding_years_professional",
    "EdLevel": "education_level",
    "ICorPM": "ic_or_manager",

    # ---- Learning ----
    "LearnCode": "learning_methods",
    "LearnCodeOnline": "learning_sources",
    "LearnCodeCoursesCert": "cert_learning_sources",

    # ---- Tech Exposure ----
    "ProfessionalTech": "tech_count",
    "ToolsTech": "tools_used",
    "Platform": "platforms_used",
    "MiscTech": "misc_tech_used",
    "OfficeStackAsync": "collab_tools_used",
    "VCInteraction": "version_control_interaction",
    "OpSys": "primary_os",

    # ---- StackOverflow ----
    "SOVisitFreq": "so_frequency",
    "SOPartFreq": "so_participation",
    "SOComm": "so_community_member",
    "NEWSOSites": "so_sites_used",

    # ---- Purchase ----
    "PurchaseInfluence": "PurchaseInfluence",
    "BuyNewTool": "BuyNewTool",

    # ---- Onboarding / Culture ----
    "Onboarding": "onboarding_speed",
    "TrueFalse_1": "supports_onboarding",
    "TrueFalse_2": "employer_learning_support",
    "TrueFalse_3": "employer_learning_time",

    # ---- Identity ----
    "MainBranch": "developer_status"
}

# ---------------------------------------------------------------
# 6.2APPLY RENAME MAP
# ---------------------------------------------------------------
df_clean.rename(columns=rename_map_2022, inplace=True)

print("Columns renamed successfully. Total columns:", df_clean.shape[1])

# ---------------------------------------------------------------
# 7. SAVE FINAL MODELLING DATASET
# ---------------------------------------------------------------
df_clean.to_csv("devx_2022_modelling_Dataset.csv", index=False)
print("✔ Saved final cleaned + imputed dataset: devx_2022_modelling_Dataset.csv")
```
