```python
import matplotlib.pyplot as plt

# Number of clusters
ks = [2,3,4,5,6,7]

# BIC values from printed output
bic_data = {
    "2022": [-3331097.78, -3419854.34, -3543931.47, -3681643.94, -3793731.50, -3806359.47],
    "2024": [-3272984.19, -3501787.38, -3604840.11, -3648051.08, -3664641.18, -3676655.19],
    "2025": [-2296696.06, -2423508.67, -2542349.91, -2628932.68, -2628696.06, -2649513.29]
}

markers = {
    "2022": "o",
    "2024": "s",
    "2025": "D"
}

colors = {
    "2022": "#1f77b4",
    "2024": "#2ca02c",
    "2025": "#d62728"
}

plt.figure(figsize=(7,5))

for year in bic_data:
    plt.plot(
        ks,
        bic_data[year],
        marker=markers[year],
        color=colors[year],
        linewidth=2,
        markersize=7,
        label=year
    )

plt.xlabel("Number of Profiles (k)", fontsize=11)
plt.ylabel("Bayesian Information Criterion (BIC)", fontsize=11)

plt.title("BIC Model Selection for Latent Profile Analysis Across Survey Years", fontsize=12)

plt.grid(axis="y", linestyle="--", alpha=0.4)

plt.legend(title="Dataset Year")

plt.tight_layout()

plt.savefig("combined_BIC_curve.png", dpi=300)

plt.show()
```
